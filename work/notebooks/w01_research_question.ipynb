### Research Question

Can machine learning prioritize existing web pages for content review more effectively than a transparent, hand-written rule, using historical search performance, content freshness, and engagement signals?

### Unit of Analysis

One web page. Each observation represents a page with features such as impressions, average search position, click-through rate (CTR), content age, and days since its last update.

### Decision

Which pages should the content team review first when editorial resources are limited?

### Model Output

A ranked list of pages with:

- A predicted probability of declining search performance.
- A priority ranking for editorial review.
- An explanation of the main factors contributing to the recommendation.

### Action

An SEO specialist or content editor examines the highest-priority pages, determines whether changes are appropriate, and may update outdated information, improve content quality, or optimize search presentation.

The model recommends which pages to investigate; it does not automatically modify content.

### Cost of Incorrect Recommendations

**False positive:** A healthy page is incorrectly prioritized for review. This wastes editorial time and may result in unnecessary changes.

**False negative:** A genuinely declining page receives low priority. The team may miss an opportunity to investigate deteriorating performance.

The relative costs depend on editorial capacity, review effort, and the potential business impact of each page.

### Why Machine Learning?

Simple rules rely on fixed thresholds and may overlook interactions between search visibility, engagement, and content freshness.

Machine learning can combine multiple signals, estimate patterns associated with declining performance, and rank pages by predicted risk.

However, predicting decline does not establish that refreshing a page will improve traffic. The system should be evaluated using held-out data and should support human decision-making rather than replace it.

### Dataset Exploration - Supporting Evidence

I analyzed the 30,000-page anonymized FlyRank starter dataset to assess whether content review prioritization is a meaningful research problem.

**Finding 1 — Declining performance is common.**

Of the 30,000 pages, 16,262 (54.2%) are classified as declining. This suggests that reviewing every declining page manually could require substantial editorial resources, making prioritization potentially valuable.

**Finding 2 — Search visibility differs across performance groups.**

Declining pages received approximately 4,919 average impressions over 90 days, compared with 9,214 for stable pages. This descriptive difference motivates further investigation into whether search visibility can help prioritize review.

**Finding 3 — Content freshness alone may not be informative enough.**

The median days since last update were 20 for declining pages and 22 for stable pages. These similar values suggest that a simple rule based only on update recency may struggle to distinguish these groups.

**Why this supports my chosen lane**

The dataset contains meaningful variation in performance and multiple observable page-level signals. A machine learning approach could investigate relationships among these features and compare its review prioritization against transparent baseline rules.

These findings are descriptive and do not establish that updating a page will cause its search performance to improve.
